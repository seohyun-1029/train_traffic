파일 읽기 및, 데이터 타입 보정 간단하게.

```python
#파일읽어오기
df_avg_traffic = pd.read_csv('avg_traffic_day.csv', encoding='cp949')       # 요일별 평균혼잡도
df_korail = pd.read_csv('korail_traffic.csv', encoding='cp949')             # 코레일 혼잡도 샘플
df_train_traffic = pd.read_csv('train_traffic.csv', encoding='cp949')       # 머신러닝용/조회용 상세 혼잡도
df_timetable = pd.read_csv('train_starttime.csv', encoding='cp949')         # 열차운행시각표
df_time = pd.read_csv('train_travletime_station.csv', encoding='cp949')     # 역간거리 소요시간

#공백제거
df_avg_traffic.columns = df_avg_traffic.columns.str.strip()
df_train_traffic.columns = df_train_traffic.columns.str.strip()
df_time.columns = df_time.columns.str.strip()

#타입보정 -> 숫자로 보정함
df_train_traffic['혼잡도'] = pd.to_numeric(df_train_traffic['혼잡도'], errors='coerce').fillna(0)

#30분 단위 보정
def format_time_to_30min(time_str):
    time_str = time_str.replace(" ", "")
    try:
        dt = datetime.strptime(time_str, "%H시%M분")
        if dt.minute < 30:
            return f"{dt.hour}시00분"
        else:
            return f"{dt.hour}시30분"
    except ValueError:
        return time_str

print("-> 데이터셋 준비 완료!\n")
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
from datetime import datetime, timedelta
import numpy as np

#사전준비 - 데이터셋 준비
raw_stations = df_train_traffic['출발역'].dropna().unique().tolist()
available_stations = sorted([str(st).strip() for st in raw_stations])

print("="*20 + " 맞춤형 이동 경로 혼잡도 탐색 " + "="*20)

#초기 정보 설정
day_type = input("요일을 입력하세요 (평일/토요일/일요일 (월-금요일은 평일로 입력)): ").strip()
start_time_input = input("출발 시간을 입력하세요 (예: 08시30분 또는 10시): ").strip().replace(" ", "")

if "시" in start_time_input and "분" not in start_time_input:
    start_time_input += "00분"
if len(start_time_input.split("시")[0]) == 1:
    start_time_input = "0" + start_time_input

try:
    current_time_dt = datetime.strptime(start_time_input, "%H시%M분")
    print(f"출발 설정 완료: [ {start_time_input} ]\n")
except ValueError:
    print(f"시간 형식 오류")
    current_time_dt = datetime.strptime("09시00분", "%H시%M분")

#경로 기록 리스트
route_history = []

start_station_text = widgets.Text(description="출발역:", placeholder="예: 합정, 서울역")
next_station_text = widgets.Text(description="도착역:", placeholder="예: 연신내, 강남")
action_button = widgets.Button(description="1. 호선 조회하기", button_style='success')
reset_button = widgets.Button(description="처음부터 다시", button_style='danger')

#동적 UI 출력
line_buttons_container = widgets.HBox([])
status_output = widgets.Output()
report_output = widgets.Output()

display(widgets.VBox([start_station_text, next_station_text, widgets.HBox([action_button, reset_button])]))
display(line_buttons_container)
display(status_output)
display(report_output)


def on_action_clicked(b):
    """출발역과 도착역을 기반으로 환승/탑승 가능한 호선 조회"""
    with status_output:
        clear_output()

        station = start_station_text.value.strip()
        next_station = next_station_text.value.strip()

        if not station or not next_station:
            print("출발역과 도착역을 모두 입력해 주세요!")
            return

        #글자 예외 보정
        search_station = station[:-1] if station.endswith('역') and len(station) > 2 else station

        if search_station not in available_stations:
            print(f"⚠️ '{station}'역은 데이터셋에 없습니다. 이름을 다시 확인해 주세요.")
            return

        #탑승 가능한 호선 찾기
        available_lines = df_train_traffic[df_train_traffic['출발역'] == search_station]['호선'].unique().tolist()
        available_lines.sort()

        print(f"👉 [{search_station}역]에서 탑승할 호선 버튼을 클릭하면 구간이 저장됩니다:")

        #호선 클릭 버튼들 동적 생성
        buttons = []
        for line in available_lines:
            btn = widgets.Button(description=str(line), button_style='info', layout=widgets.Layout(margin='3px'))

            #버튼 클릭 시 클로저 함수 생성
            def make_callback(sel_line):
                def on_line_clicked(line_b):
                    global current_time_dt

                    #도착역 '역' 글자 보정
                    dest_station = next_station[:-1] if next_station.endswith('역') and len(next_station) > 2 else next_station

                    #시간대 float 변환 매칭
                    time_float_val = current_time_dt.hour + (current_time_dt.minute / 60.0)
                    time_float_matched = round(time_float_val * 2) / 2
                    if time_float_matched < 5.5: time_float_matched = 5.5
                    if time_float_matched > 24.0: time_float_matched = 0.5

                    #혼잡도 데이터 조회
                    segment_df = df_train_traffic[
                        (df_train_traffic['출발역'] == search_station) &
                        (df_train_traffic['호선'] == sel_line) &
                        (df_train_traffic['요일구분'] == day_type) &
                        (df_train_traffic['time_float'] == time_float_matched)
                    ]

                    congestion = segment_df['혼잡도'].mean() if not segment_df.empty else 35.0
                    if np.isnan(congestion): congestion = 35.0

                    #이동 소요 시간 누적 (5분)
                    current_time_dt += timedelta(minutes=5)

                    #기록 저장
                    route_history.append({
                        '출발': search_station, '호선': sel_line, '도착': dest_station, '혼잡도': congestion
                    })

                    #상태 새로고침
                    with status_output:
                        clear_output()
                        print(f"구간 저장 완료: {search_station}역 ➔ ({sel_line}) ➔ {dest_station}역 [현재 누적시간: {current_time_dt.strftime('%H:%M')}]")
                        print("연속해서 다음 경로를 입력(환승)하시거나, 완료되었다면 아래 리포트를 확인하세요.")

                    # 결과 리포트 실시간 업데이트 출력
                    with report_output:
                        clear_output()
                        print("\n" + "="*15 + " 현재까지 누적된 이동 경로 " + "="*15)
                        total_cong = 0
                        for idx, r in enumerate(route_history, 1):
                            print(f" [{idx}구간] {r['출발']}역 ➔ ({r['호선']}) ➔ {r['도착']}역 | 혼잡도: {r['혼잡도']:.1f}")
                            total_cong += r['혼잡도']

                        avg_route_cong = total_cong / len(route_history)
                        print("-" * 65)
                        print(f"총 이동 구간: {len(route_history)}개 구간")
                        print(f"전체 경로 평균 혼잡도: {avg_route_cong:.1f}%")
                        print("-" * 65)

                        # [핵심] 평균 혼잡도 기준별 맞춤형 알림 가이드 라인 출력
                        print("📢 [종합 혼잡도 안내 브리핑]")
                        if avg_route_cong < 25:
                            print("✨ 🟢 [상태: 매우 쾌적] 현재 이동 경로는 매우 여유롭습니다. 무리 없이 편안한 이동이 가능합니다.")
                        elif avg_route_cong < 35:
                            print("⚠️ 🟡 [상태: 살짝 혼잡] 열차 내 서 있는 승객이 생기기 시작합니다. 안전 손잡이를 잡아주세요.")
                        elif avg_route_cong < 45:
                            print("🚨 🟠 [상태: 혼잡 주의] 열차 내부 밀도가 높습니다. 소요 시간이 비슷하다면 환승 우회 경로를 추천합니다.")
                        else:
                            print("🔥 🔴 [상태: 혼잡 폭발] 현재 지옥철 피크 타임 구간입니다! 가능한 경우 출발 시간을 30분 늦추는 것을 강력 권장합니다.")
                        print("=" * 65)

                    # 다음 입력을 위해 텍스트 박스 자동 세팅 (직전에 입력한 도착역이 출발역이 되도록)
                    start_station_text.value = next_station
                    next_station_text.value = ""
                    line_buttons_container.children = [] #호선 버튼 초기화

                return on_line_clicked

            btn.on_click(make_callback(str(line)))
            buttons.append(btn)

        line_buttons_container.children = buttons

def on_reset_clicked(b):
    """초기화 버튼 동작"""
    global route_history
    route_history = []
    start_station_text.value = ""
    next_station_text.value = ""
    line_buttons_container.children = []
    with status_output: clear_output(); print("모든 경로 기록이 초기화되었습니다.")
    with report_output: clear_output()

#버튼이벤트연결
action_button.on_click(on_action_clicked)
reset_button.on_click(on_reset_clicked)
```
